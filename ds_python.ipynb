# **2. Array Attributes**

```python
arr=np.ones((2,2))
print(arr.ndim)
print(arr.shape)
print(arr.size)
print(arr.dtype)
```

# **3. Reshaping Arrays**

```python
arr = np.arange(9)
arr
```

```python
arr2 = arr.reshape(3, 3)
print(arr2)
```

```python
flat = arr2.flatten() # copy of original
print(flat)
```

```python
rav = arr2.ravel() # not a copy. change here changes original too.
print(rav)
```

```python
print(arr2.T)
```

# **4. Indexing & Slicing**

```python
arr[0]
arr[1:4]
arr[:, 0]
arr[1:3, 2:4]
arr[arr > 5] # boolean filter
```

# **5. Math Operations**

```python
np.add(a, b)
np.subtract(a, b)
np.multiply(a, b)
np.divide(a, b)
np.power(a, 2)

np.sqrt(arr)
np.exp(arr)
np.log(arr)
```

# **6. Aggregations**

```python
np.sum(arr)
np.mean(arr)
np.std(arr)
np.var(arr)
np.min(arr)
np.max(arr)
np.argmin(arr) # index of max val
np.argmax(arr)
np.cumsum(arr) # cumulative sum
```

# **7. Broadcasting**

```python
arr = np.array([1, 2, 3])
print(arr + 5)
print(arr + [1,2,3])
print(arr * [1,2,3])
```

# **8. Stacking & Concatenation**

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

np.concatenate([a, b]) # joins along an existing array
```

```python
a = np.array([[1, 2],
              [3, 4]])

b = np.array([[5, 6],
              [7, 8]])

np.concatenate([a, b]) # default axis=0
```

```python
a = np.array([[1, 2],
              [3, 4]])

b = np.array([[5, 6],
              [7, 8]])

np.concatenate([a, b],axis=1)
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

np.vstack([a, b])
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

np.hstack([a, b])
```

```python
np.unique(arr) # removes duplicates
```
